```python
total_score = 0 # 合計点数
num = 0 # 人数

while True:
    # 点数の入力
    score = input("点数を入力してください：")
    # 点数の表示
    print("点数を入力してください：" + score)
    score = int(score) # strからint
    # -1が入力されたら
    if score == -1:
        break # 処理終了
    total_score += score # 入力された点数を合計点数に加算
    num += 1 # 人数カウント
    
total_score /= num # 平均点の計算
#結果表示
print(f"{num}人のテストの平均点は{round(total_score, 1)}点です")
```
