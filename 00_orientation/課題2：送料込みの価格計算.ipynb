```python
# 税抜価格の入力
price = input("税抜価格を入力してください：")
# 税抜価格の表示
print("税抜価格を入力してください：" + price)
price = float(price) # strからfloat
price *= 1.1 # 税込価格の計算(税率10％)
price = round(price) # 小数点四捨五入

# 税込価格が2000円以上の場合
if price >= 2000:
    print("送料は無料です")
# 税込価格が2000円未満の場合
elif price < 2000:
    print("送料として350円かかります")
    price += 350 # 送料加算

#送料込みの価格表示
print(f"送料込みの価格は{price}です。")
```
