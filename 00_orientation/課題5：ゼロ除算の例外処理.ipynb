```python
try:
    # 割られる数の入力、表示
    a = int(input("割られる数を入力してください："))
    print("割られる数を入力してください：", a)
    # 割る数の入力、表示
    b = int(input("割る数を入力してください："))
    print("割る数を入力してください：", b)
    c = a / b # 計算
    # 計算式の表示
    print(f"{a} ÷ {b} = {c}")
# 文字列入力エラー
except ValueError:
    print("エラー：数値を入力してください")
# 0入力エラー
except ZeroDivisionError:
    print("エラー：0で割り算しないでください")
finally:
    print("処理を終了します")
```
