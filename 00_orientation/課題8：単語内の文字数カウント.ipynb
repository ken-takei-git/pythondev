```python
# 英単語辞書の作成
eng_dict = []
alphabet = {'a':0, 'b':0, 'c':0, 'd':0, 'e':0, 'f':0, 'g':0, 'h':0, 'i':0, 'j':0, 'k':0, 'l':0,'m':0,
            'n':0, 'o':0, 'p':0, 'q':0, 'r':0, 's':0, 't':0, 'u':0, 'v':0, 'w':0, 'x':0, 'y':0, 'z':0}

# 英単語の登録
while True:
    str = "" # 初期化
    # 英単語の入力、表示
    str = input("英単語を入力してください：")
    print("英単語を入力してください：" + str)
    # 未入力の場合、処理終了
    if not str:
        break
    # 入力された英単語を構成する文字を1文字ずつ確認して、登場した数を加算    
    str_len = len(str) # 入力された英単語の長さを代入
    # 英単語の長さ分、繰り返し処理
    for i in range(0, str_len):
        alphabet[str[i]] += 1 # 対象英単語のキーをインクリメント
    eng_dict.append(str) # 英単語をリストに登録
# アルファベット順に並べ替え
eng_dict.sort()
```

```python
# 入力された単語内の文字数を表示

eng_search = [] # 英単語検索用のリスト
eng_search = list(alphabet.keys()) # 英単語コピー(リストの作成)
# 入力した英単語の一覧表示
print("入力した英単語：", eng_dict)
# アルファベット数分、繰り返し処理(abc順)
for i in range(0, 26):
    # 対象のアルファベットが1つ以上ある場合
    if alphabet[eng_search[i]] > 0 :
        # 英単語が何個あるか表示
        print(f"{eng_search[i]}が{alphabet[eng_search[i]]}個ありました")
```
