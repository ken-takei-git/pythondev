```python
#@title 九九のリストを作成する
list_add = [] # 初期化
for i in range(1, 10):
    list_inner = [] # 初期化
    for y in range(1, 10):
        list_inner.append(i * y) #九九の値、書き込み
    list_add.append(list_inner) # n段を配列ごとコピーする
```

```python
#@title 九九の表を表示する
for i in range(0, 9):
    for y in range(0, 9):
        # n段の表示
        print(f"{(list_add[i][y]):3d}", end="")
    print("") # 改行
```
