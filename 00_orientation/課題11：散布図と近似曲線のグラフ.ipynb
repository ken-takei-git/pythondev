```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([62, 95, 83, 100, 72, 71, 55, 85, 67, 75])   # english
y = np.array([62, 80, 85, 95, 58, 76, 52, 84, 67, 71])    # math

# 近似直線の式の a と b が入ったndarray 形式のデータを得る
p = np.polyfit(x, y, 1)
# 一次関数の式のオブジェクトを生成する
f = np.poly1d(p)

# 散布図と近似直線を描く
plt.xlabel("english") # x軸title
plt.ylabel("math")    # y軸title
plt.scatter(x, y, color="blue", label="score") # 散布図
plt.plot(x, f(x), color="red", label="lsm")    # 近似直線
plt.legend(loc="upper left") # 凡例を表示
plt.show() # 余分なテキスト削除
```
