```python
# +-----------------------------+
# ベースボールチームクラス        +
# 勝率の算出と勝敗情報表示を行う　 +
# +-----------------------------+
class Baseball_Team:
    name = ""
    win = 0
    lose = 0
    drow = 0

    # 勝率算出
    def calc_win_rate(self, win, lose):
        return win / (win + lose)

    # 勝敗情報表示
    def show_team_result(self, name, win, lose, draw):
        rate = self.calc_win_rate(win, lose)
        # 結果表示
        print(f"{name:<9}{win:>3}{lose:>5}{draw:>5}{rate:>6.3f}")

# インスタンス生成
archers = Baseball_Team()
meteors = Baseball_Team()
cheetahs = Baseball_Team()
sharks = Baseball_Team()
pythons = Baseball_Team()
owls = Baseball_Team()

# 見出し表示
print("team     win lose draw rate")
# チームごとの結果表示
archers.show_team_result("Archers", 77, 64, 2)
meteors.show_team_result("Meteors", 71, 69, 3)
cheetahs.show_team_result("Cheetahs", 69, 68, 6)
sharks.show_team_result("Sharks", 70, 70, 3)
pythons.show_team_result("Pythons", 68, 73, 2)
owls.show_team_result("Owls", 59, 82, 2)
```
